```python
import os
import ale_py
import gymnasium as gym
gym.register_envs(ale_py)  # Always run this first in Colab

from stable_baselines3 import DQN
from stable_baselines3.common.env_util import make_atari_env
from stable_baselines3.common.vec_env import VecFrameStack
from stable_baselines3.common.callbacks import EvalCallback, CheckpointCallback
```

```python
# ── EXPERIMENT SETTINGS ─────────────────────────────────────
# Change ONLY this block for each of your 10 experiments
EXP_NAME   = "exp1_baseline"   # change this per run e.g. exp2_lr_high

LEARNING_RATE     = 1e-4   # YOUR FOCUS: vary this across experiments
GAMMA             = 0.99
BATCH_SIZE        = 32
BUFFER_SIZE       = 100_000
EXPLORATION_FRAC  = 0.1    # epsilon_decay
EXPLORATION_INIT  = 1.0    # epsilon_start
EXPLORATION_FINAL = 0.01   # epsilon_end
TOTAL_STEPS       = 150_000  # keep low on free Colab (100k–200k)

# ── FIXED SETTINGS (don't change these) ──────────────────────
ENV_ID     = "ALE/Breakout-v5"
POLICY     = "CnnPolicy"
DRIVE_PATH = f"/content/drive/MyDrive/atari-dqn/models/{EXP_NAME}"

os.makedirs(DRIVE_PATH, exist_ok=True)
print(f"Experiment: {EXP_NAME}")
print(f"lr={LEARNING_RATE} | gamma={GAMMA} | batch={BATCH_SIZE} | steps={TOTAL_STEPS}")
```

```python
# Training environment
vec_env = make_atari_env(ENV_ID, n_envs=1, seed=42)
vec_env = VecFrameStack(vec_env, n_stack=4)

# Evaluation environment (separate — never mix with training)
eval_env = make_atari_env(ENV_ID, n_envs=1, seed=0)
eval_env = VecFrameStack(eval_env, n_stack=4)

print("✅ Environments ready")
```

```python
model = DQN(
    policy                   = POLICY,
    env                      = vec_env,
    learning_rate            = LEARNING_RATE,
    buffer_size              = BUFFER_SIZE,
    learning_starts          = 10_000,
    batch_size               = BATCH_SIZE,
    gamma                    = GAMMA,
    train_freq               = 4,
    target_update_interval   = 1000,
    exploration_fraction     = EXPLORATION_FRAC,
    exploration_initial_eps  = EXPLORATION_INIT,
    exploration_final_eps    = EXPLORATION_FINAL,
    verbose                  = 1,
    tensorboard_log          = f"{DRIVE_PATH}/tb_logs/"
)
print("✅ DQN agent ready")
```

```python
# Saves the best model automatically during eval
eval_callback = EvalCallback(
    eval_env,
    best_model_save_path = f"{DRIVE_PATH}/best/",
    log_path             = f"{DRIVE_PATH}/eval_logs/",
    eval_freq            = 10_000,
    n_eval_episodes      = 5,
    deterministic        = True,
    render               = False,
    verbose              = 1
)

# Saves a checkpoint every 50k steps (crash protection)
checkpoint_callback = CheckpointCallback(
    save_freq   = 50_000,
    save_path   = f"{DRIVE_PATH}/checkpoints/",
    name_prefix = EXP_NAME,
    verbose     = 1
)

print("✅ Callbacks ready")
```

```python
from stable_baselines3.common.callbacks import CallbackList

print(f"\n🚀 Starting training: {EXP_NAME}")
print(f"   Steps: {TOTAL_STEPS:,} | lr: {LEARNING_RATE} | gamma: {GAMMA}\n")

model.learn(
    total_timesteps = TOTAL_STEPS,
    callback        = CallbackList([eval_callback, checkpoint_callback]),
    progress_bar    = True
)

# Save final model
model.save(f"{DRIVE_PATH}/final_model")
print(f"\n✅ Training done! Model saved to {DRIVE_PATH}/final_model.zip")

vec_env.close()
eval_env.close()
```

```python
import numpy as np

# Load eval logs to get final mean reward
eval_log = np.load(
    f"{DRIVE_PATH}/eval_logs/evaluations.npz",
    allow_pickle=True
)
mean_rewards = eval_log['results'].mean(axis=1)
best_reward  = mean_rewards.max()
final_reward = mean_rewards[-1]

print(f"Experiment: {EXP_NAME}")
print(f"Best mean reward:  {best_reward:.2f}")
print(f"Final mean reward: {final_reward:.2f}")
print(f"lr={LEARNING_RATE} | gamma={GAMMA} | batch={BATCH_SIZE} | steps={TOTAL_STEPS}")
```
